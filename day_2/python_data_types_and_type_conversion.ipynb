# Python Data Types

Python's built-in data types, examples, and mutable vs immutable types.

## 1. Numeric Types

Python has three built-in numeric types: `int`, `float`, and `complex`.

### `int`

Whole numbers. Python integers support arbitrary precision.

````python
age = 30
count = -10
big_number = 12345678901234567890

print(age)
print(type(age))
````

### `float`

Numbers containing a decimal part. Scientific notation is also supported.

````python
price = 99.99
temperature = -10.5
scientific = 1.5e3

print(price)
print(scientific)
print(type(price))
````

### `complex`

Complex numbers contain a real and imaginary component. Python uses `j` for the imaginary component.

````python
z = 3 + 4j

print(z)
print(z.real)
print(z.imag)
print(type(z))
````

## 2. Boolean Type

The `bool` type has two values: `True` and `False`.

````python
is_logged_in = True
is_admin = False

print(is_logged_in)
print(type(is_logged_in))
print(isinstance(True, int))
````

## 3. Text Type

### `str`

Strings represent text and are immutable.

````python
name = "Somnath"
message = 'Hello World'

print(name)
print(type(name))

# Formatted string
age = 30
print(f"My name is {name} and I am {age} years old.")
````

## 4. Sequence Types

Python's main built-in sequence types are `list`, `tuple`, and `range`.

### `list`

An ordered, mutable collection. Lists can contain values of different types.

````python
numbers = [10, 20, 30, 40]
mixed = [10, "Hello", True, 3.14]

numbers[0] = 100
numbers.append(50)

print(numbers)
print(mixed)
print(type(numbers))
````

### `tuple`

An ordered, immutable collection.

````python
coordinates = (10, 20)
person = ("Somnath", 30, "Kolkata")

print(coordinates)
print(person)
print(type(coordinates))
````

### `range`

Represents a sequence of numbers, commonly used with loops.

````python
numbers = range(5)
print(list(numbers))

even_numbers = range(2, 10, 2)
print(list(even_numbers))
print(type(even_numbers))
````

## 5. Mapping Type

### `dict`

A dictionary stores key-value pairs and is mutable.

````python
user = {
    "name": "Somnath",
    "age": 30,
    "city": "Kolkata"
}

print(user["name"])

user["age"] = 31
user["country"] = "India"

print(user)
print(type(user))
````

## 6. Set Types

### `set`

An unordered collection of unique elements.

````python
numbers = {1, 2, 2, 3, 3}
print(numbers)
print(type(numbers))

a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
````

### `frozenset`

An immutable set.

````python
numbers = frozenset([1, 2, 3])

print(numbers)
print(type(numbers))
````

## 7. Binary Types

Python provides `bytes`, `bytearray`, and `memoryview` for working with binary data.

### `bytes`

An immutable sequence of bytes.

````python
data = b"Hello"
encoded = "Hello".encode()

print(data)
print(encoded)
print(type(data))
````

### `bytearray`

A mutable sequence of bytes.

````python
data = bytearray(b"Hello")
data[0] = 74  # ASCII value for 'J'

print(data)
print(type(data))
````

### `memoryview`

A memory view provides access to the data of a binary object without copying it.

````python
data = bytearray(b"Hello")
view = memoryview(data)

print(view)
print(type(view))
print(view[0])
````

## 8. None Type

### `NoneType`

`None` represents the absence of a value. Functions that don't explicitly return a value return `None`.

````python
result = None

print(result)
print(type(result))

def test():
    pass

print(test())
````

# Python Type Conversion

**Type conversion** in Python means converting a value from one data type to another.

Python provides built-in functions such as `int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`, and `dict()` for type conversion.

---

## 1. Implicit Type Conversion

Python automatically converts one type to another when it is safe to do so.

````python
x = 10        # int
y = 2.5       # float

result = x + y

print(result)        # 12.5
print(type(result))  # <class 'float'>
````

Here:

```text
int + float → float
```

Example

````python
x = 10
y = 3

result = x / y

print(result)
print(type(result))
````

Output:

```text
3.3333333333333335
<class 'float'>
```

Common implicit conversions

| From    | To        | Example            |
| ------- | --------- | ------------------ |
| `int`   | `float`   | `10 → 10.0`        |
| `int`   | `complex` | `10 → (10+0j)`     |
| `float` | `complex` | `10.5 → (10.5+0j)` |

## 2. Explicit Type Conversion

When you explicitly tell Python to convert a value, it is called **explicit type conversion** or **type casting**.

````python
x = "100"
y = int(x)

print(y)
print(type(y))
````

Output:

```text
100
<class 'int'>
```

### A) `int()`: converts a compatible value into an integer.

- String → Integer

````python
x = int("100")

print(x)
print(type(x))
````

Output:

```text
100
<class 'int'>
```

- Float → Integer

````python
x = int(25.9)

print(x)
````

Output:

```text
25
```

> **Note:** `int()` truncates the decimal portion. It does not round the number.

````python
int(9.99)    # 9
int(-9.99)   # -9
````

- Boolean → Integer

````python
int(True)    # 1
int(False)   # 0
````

### B) `float()`: converts a compatible value into a floating-point number.

- Integer → Float

````python
x = float(10)

print(x)
print(type(x))
````

Output:

10.0
<class 'float'>

- String → Float

````python
x = float("10.5")

print(x)
print(type(x))
````

Output:

10.5
<class 'float'>

Other examples:

````python
float(10)       # 10.0
float("25")     # 25.0
float("3.14")   # 3.14
````

### C) `str()`: converts a value into a string.

````python
x = 100

y = str(x)

print(y)
print(type(y))
````

Output:

```text
100
<class 'str'>
```

Examples

````python
str(100)        # "100"
str(10.5)       # "10.5"
str(True)       # "True"
str([1, 2, 3])  # "[1, 2, 3]"
````

Practical Example

````python
age = 25

print("My age is " + str(age))
````

### D) `bool()`: converts a value to either `True` or `False`.

````python
bool(1)        # True
bool(0)        # False

bool("hello")  # True
bool("")       # False
````

Values That Convert to `False`

The following values are generally considered **falsy**:

````python
bool(0)
bool(0.0)
bool("")
bool(None)
bool(False)
bool([])
bool(())
bool({})
bool(set())
````

All of the above return:

```text
False
```

Values That Convert to `True`

````python
bool(10)        # True
bool(-5)        # True
bool("Python")  # True
bool([1, 2])    # True
````

### Type Conversion with `input()`

One of the most important uses of type conversion is handling user input.

`input()` Always Returns a String

````python
age = input("Enter your age: ")

print(type(age))
````

If the user enters:

```text
25
```

the type is still:

```text
<class 'str'>
```

Converting Input to Integer

````python
age = int(input("Enter your age: "))

print(type(age))
````

Now the type is:

```text
<class 'int'>
```

Example: Adding Two Numbers

Without type conversion:

````python
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")

print(num1 + num2)
````

If the user enters:

```text
10
20
```

Output:

```text
1020
```

This happens because both values are strings.

With Type Conversion

````python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

result = num1 + num2

print(result)
````

Output:

```text
30
```

Invalid Type Conversion

Not every value can be converted to every type.

For example:

````python
int("hello")
````

raises:

```text
ValueError
```

Similarly:

````python
float("Python")
````

raises:

```text
ValueError
```

Handling Conversion Errors

You can use `try` and `except`:

````python
try:
    age = int(input("Enter your age: "))
    print(age)

except ValueError:
    print("Please enter a valid number")
````

Type Conversion vs Type Casting

These terms are often used interchangeably in Python.

Type Conversion

The general concept of changing one type into another.

````python
x = float(10)
````

Type Casting

Usually refers to explicitly forcing a value into another type.

````python
x = int("100")
````

For practical Python programming, you can generally treat **type conversion** and **type casting** as the same concept.

| Function    | Converts To | Example                       |

Key Points to Remember

1. **Implicit conversion** is performed automatically by Python.
2. **Explicit conversion** is performed manually using functions such as `int()`, `float()`, and `str()`.
3. `input()` **always returns a string**.
4. Use `int()` when you need an integer.
5. Use `float()` when you need a decimal number.
6. `int()` **truncates** decimals; it does not round them.
7. `bool()` converts values based on their truthiness.
8. `set()` removes duplicate values.
9. Not every value can be converted successfully.
10. Invalid conversions can raise exceptions such as `ValueError`.
