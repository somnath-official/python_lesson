# 1. Tip calculator
Write a python script to calculate tip for a bill and split the bill among the users

```python
total_bill = float(input("Enter the bill amount: "))
tip_percentage = float(input("Enter the tip percentage (10, 12 or 15): "))
total_persons = int(input("Enter the total number of persons you want to split the bill: "))

tip_amount = (total_bill * (tip_percentage / 100))
final_amount = total_bill + tip_amount
each_person_contribution = final_amount / total_persons

print(f"Your final bill is: ${final_amount}")
print(f"Each person should pay: ${each_person_contribution}")
```
