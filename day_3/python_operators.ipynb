# Python Operators

In Python, operators are symbols or keywords used to perform operations on values and variables.

### Types of Operators
---

1. Arithmetic Operators
2. Comparison Operators
3. Assignment Operators
4. Logical Operators
5. Bitwise Operators
6. Membership Operators
7. Identity Operators
8. Unary Operators

### 1. Arithmetic Operators
---

Used for mathematical calculations.

| Operator | Name           | Example   |     Result |
| -------- | -------------- | --------- | ---------: |
| `+`      | Addition       | `10 + 3`  |       `13` |
| `-`      | Subtraction    | `10 - 3`  |        `7` |
| `*`      | Multiplication | `10 * 3`  |       `30` |
| `/`      | Division       | `10 / 3`  | `3.333...` |
| `//`     | Floor Division | `10 // 3` |        `3` |
| `%`      | Modulus        | `10 % 3`  |        `1` |
| `**`     | Exponent       | `10 ** 3` |     `1000` |


### 2. Comparison Operators
---

Used to compare two values. They return True or False.

| Operator | Meaning               | Example            |
| -------- | --------------------- | ------------------ |
| `==`     | Equal                 | `5 == 5` → `True`  |
| `!=`     | Not equal             | `5 != 3` → `True`  |
| `>`      | Greater than          | `5 > 3` → `True`   |
| `<`      | Less than             | `5 < 3` → `False`  |
| `>=`     | Greater than or equal | `5 >= 5` → `True`  |
| `<=`     | Less than or equal    | `5 <= 3` → `False` |


> Important: = is assignment, while == is comparison.

````python
x = 10       # assignment
x == 10      # comparison → True
````

### 3. Assignment Operators
---

Used to assign or update values.

| Operator | Example   | Equivalent to |
| -------- | --------- | ------------- |
| `=`      | `x = 10`  | `x = 10`      |
| `+=`     | `x += 5`  | `x = x + 5`   |
| `-=`     | `x -= 5`  | `x = x - 5`   |
| `*=`     | `x *= 5`  | `x = x * 5`   |
| `/=`     | `x /= 5`  | `x = x / 5`   |
| `//=`    | `x //= 5` | `x = x // 5`  |
| `%=`     | `x %= 5`  | `x = x % 5`   |
| `**=`    | `x **= 5` | `x = x ** 5`  |

````python
x = 10
x += 5

print(x)
# 15
````

### 4. Logical Operators
---
Logical operators are used to combine multiple conditions.

Python provides three logical operators:

| Operator | Meaning                                    |
| -------- | ------------------------------------------ |
| `and`    | True if **both** conditions are True       |
| `or`     | True if **at least one** condition is True |
| `not`    | Reverses the result                        |

`and`: Both conditions must be True.

| A     | B     | `A and B` |
| ----- | ----- | --------- |
| True  | True  | True      |
| True  | False | False     |
| False | True  | False     |
| False | False | False     |

````python
age = 25
has_license = True

if age >= 18 and has_license:
    print("You can drive")
````

`or`: At least one condition must be True.

| A     | B     | `A or B` |
| ----- | ----- | -------- |
| True  | True  | True     |
| True  | False | True     |
| False | True  | True     |
| False | False | False    |

````python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend")
````

`not`: Reverses a Boolean value.

| A     | `not A` |
| ----- | ------- |
| True  | False   |
| False | True    |

````python
is_raining = False

if not is_raining:
    print("You don't need an umbrella")
````

### 5. Bitwise Operators
---

Operate on individual bits of integers.

| Operator | Name        | Example  |    |    |
| -------- | ----------- | -------- | -- | -- |
| `&`      | AND         | `5 & 3`  |    |    |
| `        | `           | OR       | `5 | 3` |
| `^`      | XOR         | `5 ^ 3`  |    |    |
| `~`      | NOT         | `~5`     |    |    |
| `<<`     | Left shift  | `5 << 1` |    |    |
| `>>`     | Right shift | `5 >> 1` |    |    |

For example:

````python
5 & 3
````

Binary:

```text
5 = 101
3 = 011
    ---
    001 = 1
```

So:

````python
5 & 3
# 1
````

### 6. Membership Operators
---

Used to check whether a value exists inside a sequence or collection.

`in`

````python
fruits = ["apple", "banana", "mango"]

"apple" in fruits
# True
````

`not in`

````python
"orange" not in fruits
# True
````

They can also be used with strings:

````python
"py" in "python"
# True
````

### 7. Identity Operators
---

Used to check whether two variables refer to the same object.

`is`

````python
a = [1, 2]
b = a

a is b
# True
````

`is not`

````python
a = [1, 2]
b = [1, 2]

a is not b
# True
````

Notice that:

````python
a == b
# True

a is b
# False
````

`==` checks value equality, while is checks object identity.

### 8. Unary Operators
---

Operators that work with a single operand.

````python
x = 10

+x      # 10
-x      # -10
~x      # bitwise NOT
not x   # False
````

### Combining Comparison and Logical Operators
---
Logical operators are commonly used with comparison operators.

````python
age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Eligible")
````

Another example:

````python
username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid credentials")
````
