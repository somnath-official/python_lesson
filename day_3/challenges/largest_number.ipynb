### Largest of Three Numbers

Take three numbers from user and print the larger number.

```python
number1 = int(input("Enter first number: "))
number2 = int(input("Enter second number: "))
number3 = int(input("Enter the third number: "))

print("Entered numbers...")
print(f"number1 -> {number1}")
print(f"number2 -> {number2}")
print(f"number3 -> {number3}")

if number1 > number2 and number1 > number3:
    print("number1 is the largest")
elif number2 > number3:
    print("number2 is the largest")
else:
    print("number3 is the largest")
```
