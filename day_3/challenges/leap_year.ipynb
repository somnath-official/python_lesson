## Check if a year is a leap year or not

#### The Rules for Leap Years

Under the modern Gregorian calendar, a year is a leap year if it meets these exact conditions:
1. Divisible by 4: Any year evenly divisible by 4 is generally a leap year (for example, 2024 and 2028).
2. The Century Exception: If the year ends in 00 (a century year like 1900 or 2100), it is not a leap year, even though it is divisible by 4.
3. The 400-Divisibility Rule: A century year is a leap year if it can be evenly divided by 400. This is why the year 2000 was a leap year, but 1700, 1800, and 1900 were not.

```python
year = int(input("Enter a year: "))

divisible_by_4 = year % 4 == 0
divisible_by_100 = year % 100 == 0
divisible_by_400 = year % 400 == 0

if divisible_by_400:
    print("The year is a leap year.")
elif divisible_by_100:
    print("The year is not a leap year.")
elif divisible_by_4:
    print("The year is a leap year.")
else:
    print("The year is not a leap year.")
```
