### Reverse a Number

Write a Python program that takes an integer as input and reverses its digits using a loop (not string slicing or built-in reversal tricks).

For example, if the input is `1234`, the output should be `4321`.

Requirements
1. Prompt the user to enter an integer.
2. Use a loop (while or for) to extract digits one at a time and build the reversed number.
3. Print the reversed number.
4. Handle negative numbers correctly — the sign should stay on the result (e.g., -123 → -321, not 321- or an error).

```python
number = int(input("Enter a number to reverse: "))
reversed = 0

while number > 0:
    reversed = (reversed * 10) + (number % 10)
    number = int(number / 10)

print(f"Reversed number is: {reversed}")
```
