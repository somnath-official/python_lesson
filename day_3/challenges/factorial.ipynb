### Factorial Calculator

Write a Python function factorial(n) that calculates the factorial of a non-negative integer n using a loop (not recursion).

The factorial of n (written n!) is the product of all positive integers less than or equal to n.

Formula:

```js
n! = n × (n-1) × (n-2) × ... × 2 × 1
```

Special case: 0! = 1 (by definition)

````python
number = int(input("Enter a number: "))
factorial = 1

n = number
while n > 1:
    factorial = factorial * n
    n -= 1

print(f"{number}! = {factorial}")
````
