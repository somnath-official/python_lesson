### Triangle Type Classifier

Write a Python program that takes three side lengths as input and determines what type of triangle they form.

A triangle can be classified as:

```text
Equilateral — all three sides are equal
Isosceles — exactly two sides are equal
Scalene — all three sides are different
```

Additionally, your program should validate whether the three given lengths can actually form a valid triangle. (Hint: recall the triangle inequality theorem — the sum of any two sides must be greater than the third side.)

Requirements
Prompt the user to enter three side lengths (as floats or integers).
First, check if the sides form a valid triangle. If not, print an appropriate message (e.g., "These sides do not form a valid triangle.") and stop.
If valid, determine and print whether it's Equilateral, Isosceles, or Scalene.
Input Format

Three positive numbers, either on separate lines or space-separated — your choice of design.

Output Format

A single line stating the triangle type, or an invalid-triangle message.


````python
side1 = int(input("Enter the first side of the triangle: "))
side2 = int(input("Enter the second side of the triangle: "))
side3 = int(input("Enter the third side of the triangle: "))

if side1 == side2 and side1 == side3 and side2 == side3:
    print("Eqilateral triangle.")
elif (side1 == side2 and side1 != side3) or (side3 == side2 and side2 != side1) or (side1 == side3 and side3 != side2):
    print("Isosceles triangle.")
else:
    print("Scalene triangle.")
````
