## Python List

A list in Python is an ordered, mutable collection used to store multiple values in a single variable.

#### 1. Creating a List

````python
numbers = [10, 20, 30, 40, 50]

names = ["Alice", "Bob", "Charlie"]

mixed = [10, "Hello", 3.14, True]
````

> A list can contain different data types.

#### 2. Accessing List Elements

Python uses zero-based indexing.

````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[0])   # Apple
print(fruits[1])   # Banana
print(fruits[3])   # Orange
````

> Negative indexing starts from the end:

````python
print(fruits[-1])  # Orange
print(fruits[-2])  # Mango
````

#### 3. Changing List Elements

Lists are mutable, meaning their elements can be changed.

````python
fruits = ["Apple", "Banana", "Mango"]

fruits[1] = "Grapes"

print(fruits)
# ['Apple', 'Grapes', 'Mango']
````

#### 4. Adding Elements

`append()`

Adds an element to the end.

````python
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
# ['Apple', 'Banana', 'Mango']
````

`insert()`

Adds an element at a specific position.

````python
fruits.insert(1, "Orange")

print(fruits)
# ['Apple', 'Orange', 'Banana', 'Mango']
````

`extend()`

Adds multiple elements.

````python
fruits.extend(["Grapes", "Watermelon"])

print(fruits)
# ['Apple', 'Orange', 'Banana', 'Mango', 'Grapes', 'Watermelon']
````

#### 5. Removing Elements

`remove()`

Removes a specific value.

````python
fruits = ["Apple", "Banana", "Mango"]

fruits.remove("Banana")

print(fruits)
# ['Apple', 'Mango']
````

`pop()`

Removes an element using its index and returns it.

````python
fruits = ["Apple", "Banana", "Mango"]

item = fruits.pop(1)

print(item)     # Banana
print(fruits)   # ['Apple', 'Mango']
````

> Without an index, it removes the last element:

````python
fruits.pop()
````

`del`

Deletes an element or a range.

````python
del fruits[0]
````

#### 6. List Slicing

Syntax:

```python
list[start:stop:step]
```

Example:

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
# [20, 30, 40]

print(numbers[:3])
# [10, 20, 30]

print(numbers[2:])
# [30, 40, 50]

print(numbers[::2])
# [10, 30, 50]

print(numbers[::-1])
# [50, 40, 30, 20, 10]
````

#### 7. Useful List Functions

````python
numbers = [10, 20, 30, 40, 50]

print(len(numbers))     # 5
print(max(numbers))     # 50
print(min(numbers))     # 10
print(sum(numbers))     # 150
````

#### 8. Useful List Methods

| Method      | Purpose                            |
| ----------- | ---------------------------------- |
| `append()`  | Add one element at the end         |
| `insert()`  | Add element at a specific position |
| `extend()`  | Add multiple elements              |
| `remove()`  | Remove a specific value            |
| `pop()`     | Remove and return an element       |
| `clear()`   | Remove all elements                |
| `index()`   | Find the index of a value          |
| `count()`   | Count occurrences                  |
| `sort()`    | Sort the list                      |
| `reverse()` | Reverse the list                   |
| `copy()`    | Create a copy                      |


Example:



````python
numbers = [30, 10, 20, 10, 40]

print(numbers.count(10))
# 2

print(numbers.index(20))
# 2

numbers.sort()
print(numbers)
# [10, 10, 20, 30, 40]

numbers.reverse()
print(numbers)
# [40, 30, 20, 10, 10]
````

#### 9. Checking Whether an Element Exists

Use in and not in.

````python
fruits = ["Apple", "Banana", "Mango"]

print("Apple" in fruits)
# True

print("Orange" in fruits)
# False

print("Orange" not in fruits)
# True
````

#### 10. Looping Through a List

````python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
````

With indexes:

````python
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

#### 11. Nested Lists

A list can contain other lists.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0])
# [1, 2, 3]

print(matrix[1][2])
# 6
````
